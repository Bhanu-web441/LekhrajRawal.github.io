```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Price distribution across Main Product Categories
category_map = {1: 'Trousers', 2: 'Skirts', 3: 'Blouses', 4: 'Sale'}
df['category_name'] = df['page 1 (main category)'].map(category_map)

sns.boxplot(ax=axes[0, 0], data=df, x='category_name', y='price', palette='Set2')
axes[0, 0].set_title('Price Distribution by Main Category')
axes[0, 0].set_xlabel('Category')
axes[0, 0].set_ylabel('Price (USD)')

# 2. Top 10 Countries by Total Clicks/Sessions
top_countries = df['country'].value_counts().head(10)
sns.barplot(ax=axes[0, 1], x=top_countries.index.astype(str), y=top_countries.values, palette='Blues_r')
axes[0, 1].set_title('Top 10 Countries by Activity (IP Category)')
axes[0, 1].set_xlabel('Country ID')
axes[0, 1].set_ylabel('Number of Clicks')

# 3. Screen Click Location vs. Frequency
sns.countplot(ax=axes[1, 0], data=df, x='location', palette='viridis')
axes[1, 0].set_title('Click Count by Screen Location (1:Top-Left to 6:Bottom-Right)')
axes[1, 0].set_xlabel('Location Code')
axes[1, 0].set_ylabel('Click Count')

# 4. Correlation Heatmap
sns.heatmap(ax=axes[1, 1], data=df[['order', 'colour', 'location', 'price', 'page']].corr(), annot=True, cmap='coolwarm', fmt=".2f")
axes[1, 1].set_title('Feature Correlation Matrix')

plt.tight_layout()
plt.show()
```
